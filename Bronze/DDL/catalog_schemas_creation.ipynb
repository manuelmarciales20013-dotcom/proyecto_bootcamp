```python
%sql
create catalog if not exists tiktok_data_eng;

create schema if not exists tiktok_data_eng.landing
comment 'esquema landing para llegada de archivos crudos';
create schema if not exists tiktok_data_eng.bronze
comment 'esquema bronze para tabla con datos sin procesar';
create schema if not exists tiktok_data_eng.silver
comment 'esquema silver para datos limpios y debidamente procesados';
create schema if not exists tiktok_data_eng.gold
comment 'esquema gold para modelado de datos y posterior consumo para BI';

create volume if not exists tiktok_data_eng.landing.archivos
comment 'volumen para llegada de archivos desde API';
```

```python
import json

ruta = "/Volumes/tiktok_data_eng/landing/archivos/tiktok_raw_20260825_011510.json"

with open(ruta, "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

print(f"Registros Raw: {len(datos)}")

tiktok_raw = spark.createDataFrame(datos)

tiktok_raw.printSchema()
```
