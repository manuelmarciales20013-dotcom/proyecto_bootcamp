# Vista Semántica: Videos por Región (`v_videos_por_region`)

Esta vista calcula la cantidad de videos generados por región geográfica (`region`) junto con sus métricas acumuladas de visualizaciones y likes:
- **Eje X**: `region` (Código de país/región).
- **Eje Y**: `cantidad_videos` (Volumen de videos producidos en esa región).
- **Métricas adicionales**: `total_vistas`, `total_likes`, `promedio_vistas`, `promedio_likes`.

**Schema:** `tiktok_data_eng.semantica`

```python
%sql
-- Creación de la vista en schema semantica
CREATE OR REPLACE VIEW tiktok_data_eng.semantica.v_videos_por_region AS
SELECT 
    s.region,
    COUNT(DISTINCT f.video_id) AS cantidad_videos,
    SUM(f.plays) AS total_vistas,
    SUM(f.likes) AS total_likes,
    ROUND(AVG(f.plays), 2) AS promedio_vistas,
    ROUND(AVG(f.likes), 2) AS promedio_likes
FROM tiktok_data_eng.gold.fct_video_metricas f
JOIN (
    SELECT DISTINCT video_id, region 
    FROM tiktok_data_eng.silver.silver_tiktok 
    WHERE region IS NOT NULL
) s ON f.video_id = s.video_id
GROUP BY s.region
ORDER BY cantidad_videos DESC;
```

```python
%sql
-- Top 15 regiones por cantidad de videos generados
SELECT * FROM tiktok_data_eng.semantica.v_videos_por_region LIMIT 15;
```
