# Vista Semántica: Top 5 Autores con más Likes (`v_top5_autores_likes`)

Esta vista identifica a los 5 creadores de contenido cuyos videos han acumulado la mayor cantidad
de likes en total dentro de la plataforma, uniendo la tabla de hechos `fct_video_metricas` con `dim_autor`.

**Schema:** `tiktok_data_eng.semantica`

```python
%sql
-- DDL: Crear schema semantica si no existe
CREATE SCHEMA IF NOT EXISTS tiktok_data_eng.semantica
COMMENT 'Capa semántica: vistas analíticas para consumo de dashboard';
```

```python
%sql
-- Creación de la vista en schema semantica
CREATE OR REPLACE VIEW tiktok_data_eng.semantica.v_top5_autores_likes AS
SELECT 
    a.autor_id,
    a.author_username,
    a.author_nickname,
    a.is_verified,
    SUM(f.likes) AS total_likes,
    SUM(f.plays) AS total_vistas,
    COUNT(f.video_id) AS total_videos,
    ROUND(AVG(f.engagement_rate), 4) AS promedio_engagement
FROM tiktok_data_eng.gold.fct_video_metricas f
JOIN tiktok_data_eng.gold.dim_autor a ON f.autor_id = a.autor_id
GROUP BY a.autor_id, a.author_username, a.author_nickname, a.is_verified
ORDER BY total_likes DESC
LIMIT 5;
```

```python
%sql
-- Consulta de validación para gráfico de barras en el dashboard
SELECT * FROM tiktok_data_eng.semantica.v_top5_autores_likes;
```
