# Vista Semántica: Nube de Palabras por Aparición de Hashtags (`v_nube_palabras_hashtags`)

Esta vista está diseñada para alimentar el componente visual de **Word Cloud (Nube de Palabras)** o Treemap en el dashboard:
- `hashtag`: El término a mostrar en la nube.
- `frecuencia_aparicion`: La cantidad de videos distintos donde aparece el hashtag (peso para el tamaño de la palabra).
- `peso_vistas` y `peso_likes`: Métricas adicionales de ponderación de color o jerarquía.

**Schema:** `tiktok_data_eng.semantica`

```python
%sql
-- Creación de la vista en schema semantica
CREATE OR REPLACE VIEW tiktok_data_eng.semantica.v_nube_palabras_hashtags AS
SELECT 
    h.hashtag,
    COUNT(DISTINCT b.video_id) AS frecuencia_aparicion,
    SUM(f.plays) AS peso_vistas,
    SUM(f.likes) AS peso_likes,
    ROUND(AVG(f.engagement_rate), 4) AS engagement_promedio
FROM tiktok_data_eng.gold.bridge_video_hashtag b
JOIN tiktok_data_eng.gold.dim_hashtag h ON b.hashtag_id = h.hashtag_id
JOIN tiktok_data_eng.gold.fct_video_metricas f ON b.video_id = f.video_id
GROUP BY h.hashtag
ORDER BY frecuencia_aparicion DESC;
```

```python
%sql
-- Consulta para visualizar los términos de mayor peso en la nube
SELECT * FROM tiktok_data_eng.semantica.v_nube_palabras_hashtags LIMIT 25;
```
