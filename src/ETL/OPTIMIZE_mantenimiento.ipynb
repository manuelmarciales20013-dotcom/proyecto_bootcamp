# Mantenimiento y Compactación Delta Lake (`OPTIMIZE`)

Este notebook ejecuta la compactación física de archivos Parquet (Bin-packing) sobre las tablas principales de las capas Silver y Gold.

### Justificación Técnica:
- **Volumetría:** El dataset contiene ~1.553 videos y ~15.700 registros puente (pocos Megabytes).
- **Evitar sobre-ingeniería:** El particionamiento en datasets de este tamaño genera el *small files problem* (cientos de archivos de pocos KB que sobrecargan el driver de Spark).
- **Estrategia:** Se consolida cada tabla en un único archivo Parquet optimizado, permitiendo lecturas secuenciales directas y aprovechamiento del **Delta Cache** en memoria del SQL Warehouse.

```python
%sql
-- Compactación capa Silver
OPTIMIZE tiktok_data_eng.silver.silver_tiktok;

-- Compactación capa Gold (Fact)
OPTIMIZE tiktok_data_eng.gold.fct_video_metricas;

-- Compactación capa Gold (Bridge)
OPTIMIZE tiktok_data_eng.gold.bridge_video_hashtag;
```

```python
%sql
-- Verificación del historial de operaciones en Delta Lake
DESCRIBE HISTORY tiktok_data_eng.silver.silver_tiktok LIMIT 5;
```
