```python
%sql
-- Esquema para la capa Bronze
create schema if not exists tiktok_data_eng.bronze
comment 'esquema bronze para tabla con datos sin procesar';

-- Tabla Bronze: copia 1:1 de Landing con todos los campos en STRING excepto fecha_ingesta
create table if not exists tiktok_data_eng.bronze.tiktok_bronze (

    -- METADATOS DE BUSQUEDA
    search_type              STRING    COMMENT 'Tipo de busqueda realizada en la API',
    search_hashtag           STRING    COMMENT 'Hashtag por el cual se hizo la busqueda',

    -- IDENTIFICACION DEL VIDEO
    video_id                 STRING    COMMENT 'Identificador unico del video en TikTok',
    description              STRING    COMMENT 'Descripcion o caption del video',
    created_at               STRING    COMMENT 'Fecha de creacion del video (ISO)',
    create_time              STRING    COMMENT 'Timestamp Unix de creacion del video',
    video_url                STRING    COMMENT 'URL directa al video',

    -- CARACTERISTICAS TECNICAS DEL VIDEO
    region                   STRING    COMMENT 'Region/pais del video',
    duration                 STRING    COMMENT 'Duracion del video en segundos',
    video_width              STRING    COMMENT 'Ancho del video en pixeles',
    video_height             STRING    COMMENT 'Alto del video en pixeles',
    ratio                    STRING    COMMENT 'Relacion de aspecto del video',
    definition               STRING    COMMENT 'Definicion del video (720p, 1080p, etc.)',
    bitrate                  STRING    COMMENT 'Bitrate del video',
    cover                    STRING    COMMENT 'URL de la imagen de portada',

    -- CLASIFICACIONES DEL VIDEO
    is_ad                    STRING    COMMENT 'Indica si el video es publicidad',
    is_photo                 STRING    COMMENT 'Indica si el contenido es foto en lugar de video',
    is_paid_content          STRING    COMMENT 'Indica si es contenido patrocinado',
    description_language     STRING    COMMENT 'Idioma detectado en la descripcion',
    duet_control             STRING    COMMENT 'Control de dueto habilitado por el autor',
    stitch_control           STRING    COMMENT 'Control de stitch habilitado por el autor',
    pinned                   STRING    COMMENT 'Indica si el video esta fijado en el perfil',

    -- ESTADISTICAS DEL VIDEO
    plays                    STRING    COMMENT 'Total de reproducciones',
    likes                    STRING    COMMENT 'Total de likes',
    comments                 STRING    COMMENT 'Total de comentarios',
    shares                   STRING    COMMENT 'Total de veces compartido',
    saves                    STRING    COMMENT 'Total de veces guardado',
    reposts                  STRING    COMMENT 'Total de reposts',

    -- IDENTIFICACION DEL AUTOR
    author_id                STRING    COMMENT 'ID interno del autor en TikTok',
    author_sec_uid           STRING    COMMENT 'ID de seguridad secundario del autor',
    author_unique_id         STRING    COMMENT 'Handle o username unico del autor',
    author_nickname          STRING    COMMENT 'Nombre visible del autor',
    author_verified          STRING    COMMENT 'Si el autor tiene verificacion oficial',
    author_verify_reason     STRING    COMMENT 'Razon textual de la verificacion',
    author_signature         STRING    COMMENT 'Biografia del autor',
    author_verification_type STRING    COMMENT 'Tipo numerico de verificacion del autor',
    author_region            STRING    COMMENT 'Region/pais del autor',

    -- ESTADISTICAS DEL AUTOR
    author_followers         STRING    COMMENT 'Total de seguidores del autor',
    author_following         STRING    COMMENT 'Total de cuentas que sigue el autor',
    author_hearts            STRING    COMMENT 'Total de likes recibidos en el perfil',
    author_videos            STRING    COMMENT 'Total de videos publicados por el autor',
    author_favoriting        STRING    COMMENT 'Total de favoritos del autor',

    -- REDES SOCIALES DEL AUTOR
    author_instagram         STRING    COMMENT 'Enlace a Instagram del autor',
    author_youtube           STRING    COMMENT 'Enlace a YouTube del autor',

    -- INFORMACION DE LA MUSICA
    music_id                 STRING    COMMENT 'ID de la musica utilizada en el video',
    music_title              STRING    COMMENT 'Titulo de la musica',
    music_author             STRING    COMMENT 'Artista o autor de la musica',
    music_is_original        STRING    COMMENT 'Indica si la musica es original del creador',
    music_duration           STRING    COMMENT 'Duracion de la musica en segundos',

    -- LISTAS
    hashtags                 STRING    COMMENT 'Lista de hashtags del video (formato texto)',
    mentions                 STRING    COMMENT 'Lista de menciones en el video (formato texto)',

    -- AUDITORIA Y LINEAJE
    fecha_ingesta            TIMESTAMP COMMENT 'Marca de tiempo del momento de ingesta para linaje y trazabilidad'

)
COMMENT 'Tabla Bronze. Copia 1:1 de Landing con schema plano. Todos los datos ingresados son STRING excepto fecha_ingesta para linaje.';
```
