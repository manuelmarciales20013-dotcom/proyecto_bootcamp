```python
%sql
-- Creación de la tabla central de hechos en la capa Gold (esquema estrella puro)
CREATE TABLE IF NOT EXISTS tiktok_data_eng.gold.fct_video_metricas (
    -- IDENTIFICADOR ÚNICO DEL HECHO
    row_hash STRING NOT NULL COMMENT 'Identificador unico del registro en hechos. Hash deterministico de video_id + | + video_url (SHA-256).',
    video_id STRING NOT NULL COMMENT 'Identificador natural unico del video en TikTok (dimension degenerada).',

    -- CLAVES DE LAS DIMENSIONES
    autor_id BIGINT NOT NULL COMMENT 'Clave foranea que referencia a dim_autor.autor_id.',
    fecha_id BIGINT NOT NULL COMMENT 'Clave foranea que referencia a dim_fecha.fecha_id.',

    -- MÉTRICAS NUMÉRICAS Y HECHOS
    plays BIGINT COMMENT 'Total de reproducciones.',
    likes BIGINT COMMENT 'Total de likes.',
    comments BIGINT COMMENT 'Total de comentarios.',
    shares BIGINT COMMENT 'Total de veces compartido.',
    saves BIGINT COMMENT 'Total de veces guardado.',
    duration INT COMMENT 'Duracion del video en segundos.',
    engagement_rate DOUBLE COMMENT 'Tasa de interaccion: (likes + comments + shares + saves) / plays.',

    -- AUDITORÍA
    _created_at TIMESTAMP COMMENT 'Marca de tiempo de insercion en la tabla de hechos.'
)
COMMENT 'Tabla central de hechos en el modelo estrella. Contiene exclusivamente las claves foraneas de dimensiones y las metricas de rendimiento cuantitativas de cada video a nivel unico (1 fila por video).';
```
