```python
%sql
-- Creación de la dimensión hashtag en la capa Gold
CREATE TABLE IF NOT EXISTS tiktok_data_eng.gold.dim_hashtag (
    hashtag_id BIGINT GENERATED ALWAYS AS IDENTITY,
    hashtag STRING NOT NULL COMMENT 'Texto del hashtag sin el simbolo #',
    longitud INT NOT NULL COMMENT 'Cantidad de caracteres del hashtag',
    _created_at TIMESTAMP COMMENT 'Marca de tiempo de insercion en la dimension'
)
COMMENT 'Dimension de hashtags en el modelo estrella. Almacena las etiquetas tematicas unicas para analizar el alcance, tendencias y tematicas clave del contenido de TikTok.';
```
