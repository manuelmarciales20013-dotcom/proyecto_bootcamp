```python
%sql
-- Creación de la dimensión autor en la capa Gold
CREATE TABLE IF NOT EXISTS tiktok_data_eng.gold.dim_autor (
    autor_id BIGINT GENERATED ALWAYS AS IDENTITY,
    author_tiktok_id STRING NOT NULL COMMENT 'ID natural unico del autor en TikTok',
    author_username STRING COMMENT 'Handle o username unico del autor (@username)',
    author_nickname STRING COMMENT 'Nombre visible o display name del autor',
    is_verified BOOLEAN COMMENT 'Indica si el autor esta verificado en TikTok',
    signature STRING COMMENT 'Biografia o descripcion del perfil del autor',
    _created_at TIMESTAMP COMMENT 'Marca de tiempo de insercion del registro para auditoria'
)
COMMENT 'Dimension de autores en el modelo estrella. Almacena el perfil y atributos descriptivos de los creadores de contenido de TikTok para analizar el desempeno, alcance e influencia en las metricas de video.';
```
