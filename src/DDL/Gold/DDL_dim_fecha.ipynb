```python
%sql
-- Creación de la dimensión fecha en la capa Gold
CREATE TABLE IF NOT EXISTS tiktok_data_eng.gold.dim_fecha (
    fecha_id BIGINT GENERATED ALWAYS AS IDENTITY,
    fecha DATE NOT NULL COMMENT 'Fecha calendario (YYYY-MM-DD)',
    anio INT NOT NULL COMMENT 'Ano numerico',
    mes INT NOT NULL COMMENT 'Numero de mes (1 a 12)',
    nombre_mes STRING NOT NULL COMMENT 'Nombre del mes en texto',
    dia INT NOT NULL COMMENT 'Dia del mes (1 a 31)',
    dia_semana INT NOT NULL COMMENT 'Dia de la semana (1 a 7)',
    nombre_dia STRING NOT NULL COMMENT 'Nombre del dia de la semana',
    trimestre INT NOT NULL COMMENT 'Trimestre del ano (1 a 4)',
    es_fin_de_semana BOOLEAN NOT NULL COMMENT 'Indica si es sabado o domingo',
    semana_anio INT NOT NULL COMMENT 'Numero de semana en el ano',
    _created_at TIMESTAMP COMMENT 'Marca de tiempo de insercion en la dimension'
)
COMMENT 'Dimension de fecha en el modelo estrella. Provee la jerarquia y atributos de calendario (ano, mes, dia, fin de semana, trimestre) para el analisis temporal de tendencias y rendimiento del contenido.';
```
