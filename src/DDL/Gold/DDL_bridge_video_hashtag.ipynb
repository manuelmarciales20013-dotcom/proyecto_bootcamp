```python
%sql
-- Creación de la tabla puente video-hashtag con ID autoincremental en la capa Gold
CREATE TABLE IF NOT EXISTS tiktok_data_eng.gold.bridge_video_hashtag (
    video_hashtag_id BIGINT GENERATED ALWAYS AS IDENTITY,
    video_id STRING NOT NULL COMMENT 'Identificador unico del video en TikTok',
    hashtag_id BIGINT NOT NULL COMMENT 'Clave foranea que referencia a dim_hashtag.hashtag_id',
    _created_at TIMESTAMP COMMENT 'Marca de tiempo de insercion en la tabla puente'
)
COMMENT 'Tabla puente (bridge) en el modelo estrella. Modela la relacion muchos a muchos (N:M) entre videos y hashtags con identificador autoincremental propio, permitiendo analizar metricas por tematicas sin generar duplicidad de reproducciones en los hechos.';
```
