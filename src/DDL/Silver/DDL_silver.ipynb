```python
%sql
CREATE SCHEMA IF NOT EXISTS tiktok_data_eng.silver
COMMENT 'Esquema silver para datos limpios, tipados y estructurados';
```

```python
%sql
CREATE TABLE IF NOT EXISTS tiktok_data_eng.silver.silver_tiktok (
    -- METADATOS DE BUSQUEDA
    search_type              STRING        COMMENT 'Tipo de busqueda que origino este video',
    search_hashtag           STRING        COMMENT 'Hashtag de busqueda que origino este video',

    -- IDENTIFICACION DEL VIDEO
    video_id                 STRING        NOT NULL COMMENT 'Identificador unico del video en TikTok. Parte de la PK natural.',
    description              STRING        COMMENT 'Descripcion del video. NULL si estaba vacio.',
    created_at               TIMESTAMP     COMMENT 'Fecha y hora de publicacion del video.',
    video_url                STRING        COMMENT 'URL directa al video en TikTok.',

    -- CARACTERISTICAS TECNICAS
    region                   STRING        COMMENT 'Region o pais del video.',
    duration                 INT           COMMENT 'Duracion del video en segundos.',
    video_width              INT           COMMENT 'Ancho del video en pixeles.',
    video_height             INT           COMMENT 'Alto del video en pixeles.',
    ratio                    STRING        COMMENT 'Relacion de aspecto del video (ej: 540p, 9:16).',

    -- CLASIFICACIONES
    is_ad                    BOOLEAN       COMMENT 'TRUE si el video es publicidad.',
    is_photo                 BOOLEAN       COMMENT 'TRUE si el contenido es una foto.',
    is_paid_content          BOOLEAN       COMMENT 'TRUE si es contenido patrocinado.',
    description_language     STRING        COMMENT 'Idioma detectado en la descripcion.',

    -- METRICAS DE INTERACCION DEL VIDEO
    plays                    BIGINT        COMMENT 'Total de reproducciones del video.',
    likes                    BIGINT        COMMENT 'Total de likes del video.',
    comments                 BIGINT        COMMENT 'Total de comentarios del video.',
    shares                   BIGINT        COMMENT 'Total de veces compartido.',
    saves                    BIGINT        COMMENT 'Total de veces guardado.',

    -- INFORMACION DEL AUTOR
    author_id                STRING        COMMENT 'ID del autor en TikTok.',
    author_unique_id         STRING        COMMENT 'Handle o username unico del autor.',
    author_nickname          STRING        COMMENT 'Nombre visible del autor.',
    author_verified          BOOLEAN       COMMENT 'TRUE si el autor tiene verificacion oficial.',
    author_signature         STRING        COMMENT 'Biografia del autor. NULL si estaba vacio.',

    -- HASHTAG INDIVIDUAL (EXPANDIDO)
    hashtag                  STRING        COMMENT 'Hashtag individual del video. Parte de la PK natural.',

    -- AUDITORIA Y LINEAJE
    fecha_ingesta            TIMESTAMP     COMMENT 'Momento en que el dato fue ingestado desde Bronze.'
)
COMMENT 'Tabla Silver. Datos limpios y tipados de TikTok. Granularidad: 1 fila por (video_id, hashtag).';
```
