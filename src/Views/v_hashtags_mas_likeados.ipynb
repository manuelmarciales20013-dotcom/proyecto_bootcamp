# Vista Semántica: Hashtags con Videos más Likeados (`v_hashtags_mas_likeados`)

Esta vista identifica los hashtags que ostentan el récord de likes en un solo video (`record_likes_un_video`)
y el total acumulado de likes, permitiendo detectar temáticas de alto impacto y viralidad.

**Schema:** `tiktok_data_eng.semantica`

```python
%sql
-- Creación de la vista en schema semantica
CREATE OR REPLACE VIEW tiktok_data_eng.semantica.v_hashtags_mas_likeados AS
SELECT 
    h.hashtag_id,
    h.hashtag,
    MAX(f.likes) AS record_likes_un_video,
    SUM(f.likes) AS total_likes_acumulados,
    ROUND(AVG(f.likes), 2) AS promedio_likes_por_video,
    COUNT(DISTINCT f.video_id) AS total_videos_con_hashtag
FROM tiktok_data_eng.gold.fct_video_metricas f
JOIN tiktok_data_eng.gold.bridge_video_hashtag b ON f.video_id = b.video_id
JOIN tiktok_data_eng.gold.dim_hashtag h ON b.hashtag_id = h.hashtag_id
GROUP BY h.hashtag_id, h.hashtag
ORDER BY record_likes_un_video DESC, total_likes_acumulados DESC;
```

```python
%sql
-- Top 10 hashtags por récord de likes
SELECT * FROM tiktok_data_eng.semantica.v_hashtags_mas_likeados LIMIT 10;
```
