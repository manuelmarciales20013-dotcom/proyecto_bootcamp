# Vista Semántica: KPIs Generales y Promedios (`v_kpi_metricas_generales`)

Esta vista consolida las métricas cuantitativas clave del dashboard (Scorecards/Tarjetas de KPI):
- **Total de videos analizados**
- **Total de vistas acumuladas**
- **Promedio de likes por video**: `AVG(likes)`
- **Promedio de vistas por video**: `AVG(plays)`
- **Promedio de comentarios, compartidos y guardados**

**Schema:** `tiktok_data_eng.semantica`

```python
%sql
-- Creación de la vista en schema semantica
CREATE OR REPLACE VIEW tiktok_data_eng.semantica.v_kpi_metricas_generales AS
SELECT 
    ROUND(AVG(plays), 2) AS promedio_vistas_por_video,
    ROUND(AVG(likes), 2) AS promedio_likes_por_video,
    ROUND(AVG(comments), 2) AS promedio_comentarios_por_video,
    ROUND(AVG(shares), 2) AS promedio_compartidos_por_video,
    ROUND(AVG(saves), 2) AS promedio_guardados_por_video,
    COUNT(DISTINCT video_id) AS total_videos,
    SUM(plays) AS total_vistas_acumuladas,
    SUM(likes) AS total_likes_acumulados
FROM tiktok_data_eng.gold.fct_video_metricas;
```

```python
%sql
-- Consulta de validación y consumo para tarjetas del dashboard
SELECT * FROM tiktok_data_eng.semantica.v_kpi_metricas_generales;
```
