# Vista Semántica: Hashtags con Videos más Vistos (`v_hashtags_mas_vistos`)

Esta vista calcula el total acumulado de reproducciones y el promedio de vistas por video
para cada hashtag, utilizando el cruce entre `fct_video_metricas`, `bridge_video_hashtag` y `dim_hashtag`.

**Schema:** `tiktok_data_eng.semantica`

```python
%sql
-- Creación de la vista en schema semantica
CREATE OR REPLACE VIEW tiktok_data_eng.semantica.v_hashtags_mas_vistos AS
SELECT 
    h.hashtag_id,
    h.hashtag,
    SUM(f.plays) AS total_vistas_acumuladas,
    ROUND(AVG(f.plays), 2) AS promedio_vistas_por_video,
    COUNT(DISTINCT f.video_id) AS total_videos_con_hashtag
FROM tiktok_data_eng.gold.fct_video_metricas f
JOIN tiktok_data_eng.gold.bridge_video_hashtag b ON f.video_id = b.video_id
JOIN tiktok_data_eng.gold.dim_hashtag h ON b.hashtag_id = h.hashtag_id
GROUP BY h.hashtag_id, h.hashtag
ORDER BY total_vistas_acumuladas DESC;
```

```python
%sql
-- Top 10 hashtags con más vistas para visualización
SELECT * FROM tiktok_data_eng.semantica.v_hashtags_mas_vistos LIMIT 10;
```
